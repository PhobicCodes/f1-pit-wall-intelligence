```python
""" 
Driver        → three-letter driver abbreviation
DriverNumber  → car number
LapTime       → total lap duration
LapNumber     → lap number for that driver
Stint         → which tyre stint they're on
PitOutTime    → when they exited pit lane
PitInTime     → when they entered pit lane
Sector1Time   → sector 1 duration
Sector2Time   → sector 2 duration
FreshTyre     → whether tyre set was fresh
Team          → constructor
Position      → driver's position
TrackStatus   → track/race-control state
Deleted       → whether lap was deleted
IsAccurate    → whether FastF1 considers timing reliable
"""
```

```python
laps.columns
```

```python
""" 
Identity / race context: tells us who/where in the race this row belongs to
  - Driver
  - DriverNumber
  - Team
  - LapNumber
  - Position
lap/sector timing: tells us how long things took
  - LapTime
  - Sector1Time
  - Sector2Time
  - Sector3Time
  - Time
  - LapStartTime
  - LapStartDate
  - Sector1SessionTime
  - Sector2SessionTime
  - Sector3SessionTime
pit/stint/tyre: strategy/tyre lifecycle
  - Stint
  - PitOutTime
  - PitInTime
  - Compound
  - TyreLife
  - FreshTyre 
speed traps: point-speed measurements
  - SpeedI1
  - SpeedI2
  - SpeedFL
  - SpeedST
Lap/status/quality information: whether the lap is valid/reliable and what track conditions applied
  - IsPersonalBest
  - TrackStatus
  - Deleted
  - DeletedReason
  - FastF1Generated
  - IsAccurate
"""
```
